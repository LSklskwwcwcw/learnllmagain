```python
import torch
import torch.nn.functional as F
import math
```

```python
def scaled_dot_product_attention(q, k, v):
  score = q @ k.transpose(-2,-1)

  dk = q.shape[-1]
  scale = math.sqrt(dk)

  scores = score / scale

  weights = F.softmax(scores,dim=-1)

  output = weights @ v
  return output,weights
```

```python
B,Tq,Tk,DK,DV = 2,4,4,8,8

q = torch.randn(B,Tq,DK)
k = torch.randn(B,Tk,DK)
v = torch.randn(B,Tk,DV)

output,weight = scaled_dot_product_attention(q,k,v)
print(output.shape)
print(weight.shape)
```

```python
B,Tq,Tk,DK,DV = 2,3,5,8,8

q = torch.randn(B,Tq,DK)
k = torch.randn(B,Tk,DK)
v = torch.randn(B,Tk,DV)

output,weight = scaled_dot_product_attention(q,k,v)
print(output.shape)
print(weight.shape)
```

```python
print(output)
```

```python
offical_output = F.scaled_dot_product_attention(q,k,v,dropout_p=0.0)
print(offical_output)
```

```python
print(offical_output == output)
```

```python
assert torch.allclose(
    output, offical_output,
    atol=1e-5, rtol=1e-5
)
```
