```python
import torch
from torch import nn
```

该类必须继承nn.Module,在此例中初始化方法中定义隐藏层和输出层，由于还需要relu，其处于第一个线性层之后，所以在前向中就需要包裹

```python
class TinyMlp(nn.Module):
  def __init__(self):
    super().__init__()

    self.hidden_layer = nn.Linear(4,8)

    self.output_layer = nn.Linear(8,2)

  def forward(self,x):

    output = self.output_layer(torch.relu(self.hidden_layer(x)))
    return output
```

```python
model = TinyMlp()

example = torch.randn(32,4)

print(example.shape)

result = model(example)
print(result.shape)
```

使用真实值和预测值相减得到loss，使用item来取出loss值

```python
criterian = nn.MSELoss()

target = torch.randn(32,2)

loss = criterian(result,target)

print(loss.item())
```

```python
loss.backward()
```

由于线性层的公式是那样，需要对矩阵做转置，所以一层是8，4而非4，8，以此类推

```python
for name,param in model.named_parameters():
  if param.grad is not None:
    print(param.grad.shape)
    print(param.grad.norm().item())
```
