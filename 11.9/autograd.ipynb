```python
import torch
x = torch.tensor([3.0],requires_grad=True)
```

```python
y=x**2 + 2*x
```

```python
y.backward()
```

```python
print(f"当 x = {x.item()} 时，f(x) 的值为: {y.item()}")
print(f"计算出的导数值 (df/dx): {x.grad.item()}")
```

```python
def numercial_grad(x,h=1e-5):
  val = x.item()

  def f(v):
    return v**2 + 2*v

  grad = (f(val+h) - f(val-h)) / (2*h)
  return grad
```

```python
result = numercial_grad(x)
print(result)
```
